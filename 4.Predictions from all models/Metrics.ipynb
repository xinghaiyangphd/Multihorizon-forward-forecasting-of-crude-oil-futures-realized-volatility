```python
import numpy as np
import pandas as pd
import math
from sklearn.metrics import mean_squared_error, r2_score, mean_absolute_error

dataframe = pd.read_excel('./WTI 1 day.xlsx', index_col=None)
dataset = dataframe.values.astype('float32')

true_values = dataset[:, 0]

error_metrics = {
    'Model': [],
    'MAE': [],
    'MAPE': [],
    'TIC': [],
    'MSE': []
}

for i in range(1, dataset.shape[1]):
    predicted_values = dataset[:, i]
    
    
    mae = mean_absolute_error(true_values, predicted_values)
    
    
    def mape(y_true, y_pred):
        return np.mean(np.abs((y_true - y_pred) / y_true)) * 100
    mape_score = mape(true_values, predicted_values)
    
    
    def tic(y_true, y_pred):
        numerator = np.sqrt(np.mean(np.square(y_pred - y_true)))
        denominator1 = np.sqrt(np.mean(np.square(y_pred)))
        denominator2 = np.sqrt(np.mean(np.square(y_true)))
        return numerator / (denominator1 + denominator2)
    tic_score = tic(true_values, predicted_values)
    
    mse = mean_squared_error(true_values, predicted_values)
    
    model_name = dataframe.columns[i] 
    error_metrics['Model'].append(model_name)
    error_metrics['MAE'].append(mae)
    error_metrics['MAPE'].append(mape_score)
    error_metrics['TIC'].append(tic_score)
    error_metrics['MSE'].append(mse)
    
    print(f'{model_name} - MAE: {mae:.4e}, MAPE: {mape_score:.4f}, TIC: {tic_score:.4f}, MSE: {mse:.4e}')

error_metrics_df1 = pd.DataFrame(error_metrics)
```
